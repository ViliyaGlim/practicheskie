1 задание

```python
n = int(input("Введите количество человек (n): "))

if n <= 0:
    print("Количество человек должно быть больше нуля.")
else:
    total_age = 0
    for i in range(1, n + 1):
        age = float(input(f"Введите возраст человека {i}: "))
        total_age += age

    average_age = total_age / n
    print(f"Средний возраст: {average_age:.2f}")
```

```python
hex_color = input("Введите цветовой код (например, a0351f): ").strip().lower()

if hex_color.startswith('#'):
    hex_color = hex_color[1:]

if len(hex_color) != 6:
    print("Некорректная длина кода. Должно быть 6 символов.")
else:
    if hex_color == "000000":
        print("Черный")
    elif hex_color == "ffffff":
        print("Белый")
    else:
        r = int(hex_color[0:2], 16)
        g = int(hex_color[2:4], 16)
        b = int(hex_color[4:6], 16)

        if r == g == b:
            print("Серый")
        else:
            max_val = max(r, g, b)

            if r == max_val and r > g and r > b:
                print("Ближе к красному")
            elif g == max_val and g > r and g > b:
                print("Ближе к зеленому")
            elif b == max_val and b > r and b > g:
                print("Ближе к синему")
            else:
                dominant = []
                if r == max_val:
                    dominant.append("красному")
                if g == max_val:
                    dominant.append("зеленому")
                if b == max_val:
                    dominant.append("синему")
                print("Ближе к", " и ".join(dominant))
```
