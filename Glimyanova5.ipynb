```python
import json

try:
    with open("students.json", "r", encoding="utf-8") as f:
        data = json.load(f)

    if not isinstance(data, list):
        raise TypeError("Данные в файле должны быть списком.")

    students_report = []
    all_grades = []

    for index, student in enumerate(data):
        if not isinstance(student, dict):
            raise TypeError(f"Элемент с индексом {index} не является словарем.")

        if "name" not in student:
            raise KeyError(f"Отсутствует обязательный ключ 'name' у элемента с индексом {index}.")
        if "grades" not in student:
            raise KeyError(f"Отсутствует обязательный ключ 'grades' у студента '{student.get('name', index)}'.")

        name = student["name"]
        grades = student["grades"]

        if not isinstance(name, str):
            raise TypeError(f"Имя студента должно быть строкой (индекс {index}).")
        if not isinstance(grades, list):
            raise TypeError(f"Оценки студента '{name}' должны быть списком.")

        for grade in grades:
            if not isinstance(grade, (int, float)):
                raise TypeError(f"Неверный тип данных в оценках у студента '{name}': найдено {type(grade).__name__}.")

        student_avg = round(sum(grades) / len(grades), 2) if grades else 0.0
        students_report.append({
            "name": name,
            "average_grade": student_avg
        })
        all_grades.extend(grades)

    overall_avg = round(sum(all_grades) / len(all_grades), 2) if all_grades else 0.0

    report = {
        "students": students_report,
        "overall_average": overall_avg
    }

    with open("report.json", "w", encoding="utf-8") as f:
        json.dump(report, f, ensure_ascii=False, indent=4)

    print(f"Отчет успешно сохранен в файл 'report.json'.")

except FileNotFoundError:
    print(f"Ошибка: Файл 'students.json' не найден.")
except json.JSONDecodeError as e:
    print(f"Ошибка: Некорректный формат JSON-файла: {e}")
except (KeyError, TypeError) as e:
    print(f"Ошибка структуры/типа данных: {e}")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```
