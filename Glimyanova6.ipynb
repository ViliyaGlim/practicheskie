```python
class Student:
    def __init__(self, name, group, tel_number, grades=None):
        self.name = name
        self.group = group
        self.__tel_number = tel_number  # Приватный атрибут
        self.grades = grades if grades is not None else []

    def add_grade(self, grade):
        """Метод добавления оценки студенту."""
        self.grades.append(grade)

    def get_average(self):
        """Метод вычисления среднего балла."""
        if not self.grades:
            return 0.0
        return round(sum(self.grades) / len(self.grades), 2)

    def info(self):
        """Метод вывода всей информации о студенте."""
        print(f"Студент: {self.name}")
        print(f"Группа: {self.group}")
        print(f"Телефон: {self.__tel_number}")
        print(f"Оценки: {self.grades}")
        print(f"Средний балл: {self.get_average()}")
        print("-" * 30)
```

Три студента:

```python
student1 = Student("Иван Иванов", "ИС-21", "+7 (999) 111-22-33", [4, 5, 4])
student2 = Student("Анна Смирнова", "ИС-21", "+7 (999) 444-55-66", [5, 5, 5])
student3 = Student("Петр Сидоров", "ИС-22", "+7 (999) 777-88-99")
```

Добавление новых оценок:

```python
student1.add_grade(5)
student2.add_grade(4)
student3.add_grade(3)
student3.add_grade(4)
```

Информация о студентах:

```python
print(student1.info())
print(student2.info())
print(student3.info())
```
